```python
import pandas as pd
import os

# 1. 设置文件夹路径
# 这里是输入文件所在的目录，也是输出文件保存的目录
folder_path = "/mnt/c/Users/Future/OneDrive/UC_project/scRNA-seq_data/06_annotation/Merge/"

# 2. 需要合并的四个文件名
file_names = [
    "immune.annotation.csv",
    "stromal.annotation.csv",
    "epithelial.annotation.csv",
    "other.annotation.csv"
]

# 3. 读取并合并
data_frames = []
for file in file_names:
    file_path = os.path.join(folder_path, file)
    print(f"正在读取: {file_path}")
    
    # 读取 CSV
    # 假设所有文件都有相同的列结构
    df = pd.read_csv(file_path)
    data_frames.append(df)

# 纵向合并 (axis=0)
# ignore_index=True 会重置索引，保证合并后的索引是连续的 0, 1, 2, ...
merged_df = pd.concat(data_frames, axis=0, ignore_index=True)

# 4. 定义输出文件路径
output_file_name = "L3_annotation.csv"  # 您可以修改这个文件名
output_path = os.path.join(folder_path, output_file_name)

# 5. 保存文件
# index=False 表示不保存行索引（0, 1, 2...）到 CSV 中
merged_df.to_csv(output_path, index=False)

print("-" * 30)
print(f"合并完成！")
print(f"合并后的文件已保存至: {output_path}")
print(f"最终数据形状: {merged_df.shape}")
```

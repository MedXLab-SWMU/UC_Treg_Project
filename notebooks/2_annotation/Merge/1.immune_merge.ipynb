```python
import pandas as pd
import os
```

```python
# 设置文件所在的文件夹路径
folder_path = "/mnt/c/Users/Future/OneDrive/UC_project/scRNA-seq_data/06_annotation/L3/immune"
```

```python
# 需要合并的五个文件名列表
file_names = [
    "L3_TNK_summary.csv",
    "L3_B_summary.csv",
    "L3_Mast_summary.csv",
    "L3_Myeloid_summary.csv",
    "L3_Plasma_summary.csv"
]

# 读取并合并
data_frames = []
for file in file_names:
    file_path = os.path.join(folder_path, file)
    # 读取CSV
    df = pd.read_csv(file_path)
    data_frames.append(df)

# 使用 concat 进行纵向合并 (axis=0)，ignore_index=True 会重置索引
merged_df = pd.concat(data_frames, axis=0, ignore_index=True)

# 打印合并后的形状以确认
print(f"合并完成，数据形状为: {merged_df.shape}")
```
